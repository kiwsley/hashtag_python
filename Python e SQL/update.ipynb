```python
import pyodbc

dados_conexao = ("Driver={SQLite3 ODBC Driver};" 
                 "Server=localhost;" 
                 "Database=chinook.db")

conexao = pyodbc.connect(dados_conexao)
print("conexão bem sucedida")
cursor = conexao.cursor()
```

```python
cursor.execute('''
UPDATE customers SET Email="kiwsley@embraer.com.br" WHERE Email="luisg@embraer.com.br" 
               ''') #executar o comando
cursor.commit()#Atualizar as informações
```

```python
cursor.close()
conexao.close() #Finalizando a conexao
```
